```python
import pandas as pd
import matplotlib.pyplot as plt

veri = pd.read_csv("student-mat.csv", sep=";")  # ayraç ; olabilir, head() ile kontrol et
print(veri.shape)
print(veri.columns.tolist())
print(veri.isna().sum().sum())
```

```python
assert veri["G3"].between(0, 20).all()
```

```python
print(veri["G3"].describe().round(2))
plt.hist(veri["G3"], bins=10, rwidth=0.85, edgecolor="black")
plt.xlabel("G3 (final notu)")
plt.ylabel("Öğrenci Sayısı")
plt.show()
```

```python
from scipy.stats import spearmanr

for s in ["studytime", "absences", "failures"]:
    r, p = spearmanr(veri[s], veri["G3"])
    print(s, "r =", round(r, 2), "p =", round(p, 4))
```

```python
print(veri.groupby("school")["G3"].mean().round(2))
print(veri.groupby("sex")["G3"].mean().round(2))
veri.boxplot(column="G3", by="studytime")
plt.show()
```
